# Week 2 – Day 3 | Task 1: Imputing and Encoding

**Concepts:** SimpleImputer, KNNImputer, OneHotEncoder, OrdinalEncoder, LabelEncoder

This notebook handles missing values, compares two numeric imputation methods, encodes categorical columns, and prints the DataFrame after every major step.

## 1. Import required libraries

```python
import pandas as pd
from sklearn.impute import SimpleImputer, KNNImputer
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, LabelEncoder
```

## 2. Create the original DataFrame with missing values

```python
df = pd.DataFrame({
    'age': [25, None, 40, 35, None],
    'income': [30000, 45000, None, 80000, 52000],
    'city': ['Chennai', 'Mumbai', None, 'Chennai', 'Delhi'],
    'size': ['S', 'L', 'M', 'S', 'L']
})

print("Original DataFrame:")
print(df)
print("\\nMissing values in each column:")
print(df.isnull().sum())
```

## 3. Fill missing values using SimpleImputer

- `age`: median strategy
- `income`: median strategy
- `city`: most-frequent strategy

The `size` column has no missing values.

```python
df_simple = df.copy()

age_imputer = SimpleImputer(strategy='median')
df_simple[['age']] = age_imputer.fit_transform(df_simple[['age']])

income_imputer = SimpleImputer(strategy='median')
df_simple[['income']] = income_imputer.fit_transform(df_simple[['income']])

city_imputer = SimpleImputer(strategy='most_frequent')
df_simple[['city']] = city_imputer.fit_transform(df_simple[['city']])

print("DataFrame after SimpleImputer:")
print(df_simple)
print("\\nMissing values after imputation:")
print(df_simple.isnull().sum())
```

## 4. Try KNNImputer on numeric columns

KNNImputer estimates a missing numeric value using nearby rows (nearest neighbors). We apply it to `age` and `income` and compare its results with median imputation.

```python
numeric_columns = ['age', 'income']

knn_imputer = KNNImputer(n_neighbors=2)
df_knn = df.copy()
df_knn[numeric_columns] = knn_imputer.fit_transform(df_knn[numeric_columns])

print("Numeric columns after KNNImputer:")
print(df_knn[numeric_columns])

print("\\nNumeric columns after SimpleImputer (median):")
print(df_simple[numeric_columns])

print("\\nComparison of imputation methods:")
comparison = pd.DataFrame({
    'SimpleImputer_median_age': df_simple['age'],
    'KNNImputer_age': df_knn['age'],
    'SimpleImputer_median_income': df_simple['income'],
    'KNNImputer_income': df_knn['income']
})
print(comparison)
```

## 5. Prepare the DataFrame for encoding

Use the SimpleImputer result as the main dataset for the remaining steps.

```python
df_encoded = df_simple.copy()

print("DataFrame before encoding:")
print(df_encoded)
```

## 6. Encode city using OneHotEncoder

OneHotEncoder creates a separate binary column for each city. `handle_unknown='ignore'` prevents errors if an unseen city appears later.

```python
onehot_encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')

city_encoded_array = onehot_encoder.fit_transform(df_encoded[['city']])
city_encoded_columns = onehot_encoder.get_feature_names_out(['city'])

city_encoded_df = pd.DataFrame(
    city_encoded_array,
    columns=city_encoded_columns,
    index=df_encoded.index
)

df_encoded = pd.concat(
    [df_encoded.drop(columns=['city']), city_encoded_df],
    axis=1
)

print("DataFrame after OneHotEncoding city:")
print(df_encoded)
```

## 7. Encode size using OrdinalEncoder

The required order is **S < M < L**, so explicitly provide that category order to the encoder.

```python
ordinal_encoder = OrdinalEncoder(categories=[['S', 'M', 'L']])

df_encoded[['size']] = ordinal_encoder.fit_transform(df_encoded[['size']])

print("DataFrame after OrdinalEncoding size (S=0, M=1, L=2):")
print(df_encoded)
```

## 8. Demonstrate LabelEncoder on a target column

The supplied dataset does not contain a target column. To demonstrate LabelEncoder, we add a small example target called `customer_type`. This is only an illustrative target, not part of the supplied input data.

```python
target = pd.Series(['Regular', 'Premium', 'Regular', 'Premium', 'Regular'],
                   name='customer_type')

label_encoder = LabelEncoder()
target_encoded = label_encoder.fit_transform(target)

target_df = pd.DataFrame({
    'customer_type': target,
    'customer_type_encoded': target_encoded
})

print("Original target and LabelEncoded target:")
print(target_df)
print("\\nClass mapping:")
for class_name, encoded_value in zip(label_encoder.classes_,
                                     label_encoder.transform(label_encoder.classes_)):
    print(f"{class_name} = {encoded_value}")
```

### Why should LabelEncoder not be used for input features?

LabelEncoder assigns integer labels to categories. For input features, these numbers can falsely suggest an order or distance between categories (for example, Chennai = 0 and Delhi = 1). Use OneHotEncoder for nominal categories and OrdinalEncoder when a meaningful order exists. LabelEncoder is generally intended for the target labels (`y`).

## 9. Final encoded DataFrame

```python
print("Final encoded feature DataFrame:")
print(df_encoded)

print("\\nFinal DataFrame information:")
print(df_encoded.info())
```

## 10. Conclusion

- `SimpleImputer` filled missing age and income values with their medians and city with its most frequent value.
- `KNNImputer` estimated missing numeric values using nearby rows; its estimates can differ from median imputation.
- `OneHotEncoder` converted city into separate binary columns.
- `OrdinalEncoder` converted size using the meaningful order S=0, M=1, L=2.
- `LabelEncoder` was demonstrated on an example target column. It should generally not be used to encode nominal input features.
